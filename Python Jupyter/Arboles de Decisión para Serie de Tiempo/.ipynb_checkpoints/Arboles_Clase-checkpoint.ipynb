```python
# Librerías
import warnings
warnings.filterwarnings("ignore", message="IProgress not found")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import pacf

from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import root_mean_squared_error, r2_score

from skforecast.recursive import ForecasterRecursive
from skforecast.direct import ForecasterDirect
from skforecast.model_selection import TimeSeriesFold, backtesting_forecaster, grid_search_forecaster
from skforecast.exceptions import IgnoredArgumentWarning
import skforecast, sklearn, shap

warnings.filterwarnings("ignore", category=IgnoredArgumentWarning)
plt.rcParams["figure.figsize"] = (10, 4)
SEMILLA = 123

print("skforecast", skforecast.__version__, "| scikit-learn", sklearn.__version__, "| shap", shap.__version__)
```

Usaremos una única función para evaluar todos los modelos, así todas las comparaciones usan las mismas métricas:

- **RECM** (raíz del error cuadrático medio): está en las mismas unidades que la serie (µg/m³). Menor es mejor.
- **R²**: proporción de variabilidad explicada. 1 es perfecto; 0 equivale a predecir siempre la media; puede ser negativo si el modelo es peor que la media.

```python
def evaluar(y_real, y_pred):
    # Devuelve la RECM y el R² de un conjunto de predicciones.
    return {"RECM": root_mean_squared_error(y_real, y_pred), "R2": r2_score(y_real, y_pred)}
```

# 1. Datos y exploración

La serie contiene la concentración diaria promedio de PM2.5 (µg/m³) en Shanghái desde el 1 de enero de 2013.

```python
pm25 = pd.read_csv("d-Shanghai-1317.txt", header=None, names=["PM25"])["PM25"]
pm25.index = pd.date_range(start="2013-01-01", periods=len(pm25), freq="D")

print(f"Observaciones: {len(pm25)}  |  desde {pm25.index[0].date()} hasta {pm25.index[-1].date()}")
pm25.head()
```

```python
pm25.plot(title="PM2.5 diario en Shanghái", ylabel="µg/m³")
plt.show()
```

## 1.1 ACF y PACF

- La **ACF** muestra la correlación entre $y_t$ y $y_{t-k}$.
- La **PACF** muestra la correlación entre $y_t$ y $y_{t-k}$ *descontando* el efecto de los rezagos intermedios.

Buscamos qué rezagos aportan información para predecir el valor de hoy.

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 6))
plot_acf(pm25, lags=400, ax=ax[0], title="ACF (400 rezagos)")
plot_pacf(pm25, lags=40, method="ywm", alpha=0.01, ax=ax[1], title="PACF (40 rezagos, bandas al 99%)")
plt.tight_layout()
plt.show()
```

Una forma objetiva de leer la PACF: un coeficiente es significativo al 99% si $|\hat\phi_{kk}| > 2.58/\sqrt{T}$. Contamos cuántos rezagos iniciales seguidos son significativos.

```python
T = len(pm25)
pacf_valores = pacf(pm25, nlags=40, method="ywm")
umbral = 2.58 / np.sqrt(T)

significativos = np.abs(pacf_valores[1:]) > umbral        # se omite el rezago 0
primer_no_signif = np.argmin(significativos) + 1           # primer rezago NO significativo
print(f"Umbral: {umbral:.3f}")
print(f"Rezagos iniciales significativos consecutivos: {primer_no_signif - 1}")
print("Rezagos significativos entre 1 y 40:", [k for k in range(1, 41) if significativos[k - 1]])
```

## 1.2 Periodograma: ¿hay ciclos?

El periodograma descompone la serie en frecuencias. Un pico en la frecuencia $f$ indica un ciclo de **periodo $1/f$ días**. Para que sea más fácil de leer, graficamos contra el **periodo** en lugar de la frecuencia.

```python
frecuencias, potencia = signal.periodogram(pm25.values)
frecuencias, potencia = frecuencias[1:], potencia[1:]      # se omite la frecuencia 0 (la media)
periodos = 1 / frecuencias

top = np.argsort(potencia)[::-1][:5]
print("Periodos (días) con mayor potencia:", np.round(periodos[top], 1))

plt.semilogx(periodos, potencia)
plt.axvline(365, color="red", ls="--", label="365 días")
plt.xlabel("Periodo (días, escala log)"); plt.ylabel("Potencia")
plt.title("Periodograma en función del periodo")
plt.legend(); plt.show()
```

**¿Qué observamos?**

- La PACF indica que el **día anterior** es, con diferencia, el rezago más informativo, con algunos rezagos cortos adicionales.
- El periodograma tiene su pico más alto en un periodo cercano a **un año**: el PM2.5 sube en invierno y baja en verano.

**Decisión de modelación.** Usaremos dos grupos de rezagos:

| Grupo | Rezagos | Qué capturan |
|---|---|---|
| Corto plazo | 1, 2, …, 10 | La inercia de los últimos días |
| Estacional | 364, 365, …, 370 | Lo que pasó hace aproximadamente un año (una semana alrededor) |

# 2. De serie a tabla: variables rezagadas

Un árbol de regresión necesita una tabla: cada **fila** es un ejemplo, las **columnas** son las variables explicativas y hay una **respuesta**. Para una serie de tiempo construimos esa tabla así:

- Respuesta: $y_t$.
- Variables explicativas: los valores pasados $y_{t-1}, y_{t-2}, \dots$ (los **rezagos**).

En `pandas`, `serie.shift(k)` desplaza la serie $k$ posiciones hacia abajo: en la fila de la fecha $t$ queda el valor de $t-k$.

**Ejemplo pequeño** con solo dos rezagos y los primeros 6 días, para ver la idea:

```python
ejemplo = pd.DataFrame({"y_t": pm25, "lag_1": pm25.shift(1), "lag_2": pm25.shift(2)}).head(6)
ejemplo
```

Las primeras filas tienen `NaN` porque no existe el pasado necesario; esas filas se descartan. Ahora construimos la tabla completa con los rezagos elegidos en la sección 1. Los nombres `lag_k` son los mismos que usa `skforecast`.

```python
LAGS = list(range(1, 11)) + list(range(364, 371))   # rezagos de corto plazo + estacionales
VENTANA = max(LAGS)                                 # días de historia necesarios para formar una fila
print("Rezagos:", LAGS)
print("Ventana:", VENTANA, "días")
```

```python
datos = pd.DataFrame(index=pm25.index)
for k in LAGS:
    datos[f"lag_{k}"] = pm25.shift(k)
datos["y"] = pm25

# Las primeras VENTANA filas quedan incompletas (tienen NaN): se eliminan
datos = datos.iloc[VENTANA:]

print(f"Filas: {len(pm25)} - {VENTANA} = {len(datos)}")
datos.head()
```

**Cada fila es un ejemplo de entrenamiento.** Por ejemplo, la primera fila dice: *"con estos 17 valores pasados, el valor de hoy fue `y`"*. El siguiente gráfico muestra de dónde salen las columnas de una fila (tomamos la primera fecha del conjunto de prueba como ilustración).

```python
fecha_obj = datos.index[992]   # una fecha cualquiera, a modo de ilustración
fechas_lag = [fecha_obj - pd.Timedelta(days=k) for k in LAGS]
corto = [f for f, k in zip(fechas_lag, LAGS) if k <= 10]
estac = [f for f, k in zip(fechas_lag, LAGS) if k >= 364]

fig, ax = plt.subplots(1, 2, figsize=(12, 3.5), sharey=True, gridspec_kw={"width_ratios": [1, 1]})
for a, fechas, color, titulo in [(ax[0], estac, "tab:orange", "Rezagos estacionales (t-370 … t-364)"),
                                 (ax[1], corto, "tab:blue", "Rezagos de corto plazo (t-10 … t-1)")]:
    ini, fin = min(fechas) - pd.Timedelta(days=5), max(fechas) + pd.Timedelta(days=5)
    pm25[ini:fin].plot(ax=a, color="lightgray")
    a.scatter(fechas, pm25[fechas], color=color, zorder=3, label="variables explicativas")
    a.set_title(titulo)
ax[1].scatter([fecha_obj], [pm25[fecha_obj]], color="green", s=80, zorder=3, label="respuesta $y_t$")
ax[1].set_xlim(right=fecha_obj + pd.Timedelta(days=3))
ax[0].set_ylabel("µg/m³")
for a in ax: a.legend(loc="upper left", fontsize=8)
fig.suptitle(f"Fila de la tabla para t = {fecha_obj.date()}")
plt.tight_layout(); plt.show()

datos.loc[[fecha_obj]]
```

# 3. División temporal: entrenamiento, validación y prueba

En series de tiempo **no se barajan los datos**: el modelo debe entrenarse con el pasado y evaluarse con el futuro, como ocurrirá en la práctica.

| Conjunto | Proporción | Uso |
|---|---|---|
| Entrenamiento | 70% | Ajustar el árbol |
| Validación | 10% | Elegir hiperparámetros (`max_depth`, poda) |
| Prueba | 20% | Evaluar **una sola vez** el modelo final |

```python
n = len(datos)
n_train = int(n * 0.70)
n_val = int(n * 0.10)
n_test = n - n_train - n_val
print(f"Total {n}  =  entrenamiento {n_train}  +  validación {n_val}  +  prueba {n_test}")

X = datos.drop(columns="y")
y = datos["y"]

X_train, y_train = X.iloc[:n_train], y.iloc[:n_train]
X_val,   y_val   = X.iloc[n_train:n_train + n_val], y.iloc[n_train:n_train + n_val]
X_test,  y_test  = X.iloc[n_train + n_val:], y.iloc[n_train + n_val:]
```

```python
ax = pm25.plot(color="black", lw=0.6, label="serie completa")
for conj, color, nombre in [(y_train, "tab:blue", "entrenamiento"), (y_val, "tab:orange", "validación"), (y_test, "tab:green", "prueba")]:
    ax.axvspan(conj.index[0], conj.index[-1], color=color, alpha=0.2, label=nombre)
ax.axvspan(pm25.index[0], X_train.index[0], color="gray", alpha=0.15, hatch="//", label=f"ventana inicial ({VENTANA} días)")
ax.set_title("División temporal de los datos"); ax.set_ylabel("µg/m³")
ax.legend(loc="upper right", fontsize=8); plt.show()
```

La primera fila de entrenamiento necesita 370 días de historia, por eso el primer año aparece como "ventana inicial": esos datos se usan **solo como rezagos**, nunca como respuesta.

# 4. Árbol de regresión construido a mano

## 4.1 ¿Cómo decide un árbol?

Un árbol de regresión divide el espacio de las variables en regiones con preguntas del tipo *"¿`lag_1` ≤ c?"*. En cada división elige la variable y el corte $c$ que **más reducen el error cuadrático**. En cada hoja, la predicción es el **promedio** de las respuestas de entrenamiento que caen en ella.

Para verlo, entrenamos árboles que **solo usan `lag_1`**: la predicción resulta ser una **función escalonada**.

```python
grid_lag1 = pd.DataFrame({"lag_1": np.linspace(X_train["lag_1"].min(), X_train["lag_1"].max(), 500)})

plt.scatter(X_train["lag_1"], y_train, s=8, alpha=0.3, color="gray", label="entrenamiento")
for prof, color in [(1, "tab:red"), (2, "tab:blue"), (4, "tab:green")]:
    arbol_1var = DecisionTreeRegressor(max_depth=prof, random_state=SEMILLA).fit(X_train[["lag_1"]], y_train)
    plt.plot(grid_lag1["lag_1"], arbol_1var.predict(grid_lag1), color=color, lw=2, label=f"max_depth={prof}")
plt.xlabel("lag_1 (PM2.5 de ayer)"); plt.ylabel("y (PM2.5 de hoy)")
plt.title("Un árbol predice con escalones: el promedio de cada región")
plt.legend(); plt.show()
```

**Observe:**
- Con más profundidad hay más escalones, y el modelo se ajusta más a los datos.
- Fuera del rango observado la predicción es **constante**: el árbol no puede predecir valores mayores al mayor promedio de sus hojas. Por eso no extrapola (ni sigue tendencias).

Ahora un árbol pequeño con **todas** las variables, para leer sus reglas:

```python
arbol_pequeno = DecisionTreeRegressor(max_depth=2, random_state=SEMILLA).fit(X_train, y_train)

plt.figure(figsize=(12, 5))
plot_tree(arbol_pequeno, feature_names=X.columns, filled=True, rounded=True, fontsize=10)
plt.title("Árbol de profundidad 2")
plt.show()
```

**Cómo leer cada nodo:** la condición de división (si se cumple, se va a la izquierda), `squared_error` = varianza de las respuestas del nodo, `samples` = número de días, `value` = promedio de PM2.5 en ese nodo, que es la predicción si el nodo es una hoja.

## 4.2 Árbol sin restricciones: sobreajuste

Si no limitamos el árbol, crece hasta que cada hoja tiene muy pocas observaciones: memoriza el entrenamiento.

```python
arbol_completo = DecisionTreeRegressor(random_state=SEMILLA).fit(X_train, y_train)

print("Profundidad:", arbol_completo.get_depth(), "| Hojas:", arbol_completo.get_n_leaves())
pd.DataFrame({"entrenamiento": evaluar(y_train, arbol_completo.predict(X_train)),
              "validación":    evaluar(y_val,   arbol_completo.predict(X_val))}).round(3)
```

El R² de entrenamiento es 1 (memorización perfecta) y en validación es negativo: el árbol completo predice **peor que la media**. Esto es **sobreajuste**. Note que **no miramos el conjunto de prueba**: se reserva para el final.

## 4.3 Elegir la profundidad con el conjunto de validación

```python
resultados_prof = []
for d in range(1, 16):
    arbol_d = DecisionTreeRegressor(max_depth=d, random_state=SEMILLA).fit(X_train, y_train)
    resultados_prof.append({"max_depth": d,
                            "RECM_entrenamiento": root_mean_squared_error(y_train, arbol_d.predict(X_train)),
                            "RECM_validación":    root_mean_squared_error(y_val, arbol_d.predict(X_val))})
resultados_prof = pd.DataFrame(resultados_prof).set_index("max_depth")

MEJOR_PROF = resultados_prof["RECM_validación"].idxmin()
print("Mejor max_depth según validación:", MEJOR_PROF)

ax = resultados_prof.plot(marker="o")
ax.axvline(MEJOR_PROF, color="red", ls="--", label=f"mínimo en validación (max_depth={MEJOR_PROF})")
ax.set_ylabel("RECM"); ax.set_title("Curva de validación"); ax.legend(); plt.show()

resultados_prof.round(3)
```

**Lectura de la curva:**
- La RECM de entrenamiento **siempre baja** al aumentar la profundidad.
- La RECM de validación baja hasta cierto punto y luego **sube**: a partir de ahí el árbol aprende ruido.
- Elegimos la profundidad con menor error en validación.

## 4.4 Alternativa: validación cruzada secuencial

En la sección 4.3 elegimos `max_depth` con **un solo bloque** de validación (124 días, de mayo a septiembre de 2016). El resultado depende de ese bloque: si cae en verano, cuando la contaminación es baja y fácil de predecir, el error parece pequeño, y la profundidad elegida puede no ser la mejor para el resto del año.

La **validación cruzada secuencial** evalúa el modelo en **varios bloques consecutivos** y promedia los errores.

**¿Por qué no la validación cruzada usual (k-fold)?** Porque baraja los datos: el modelo se entrenaría con días posteriores a los que valida, es decir, usaría el futuro para predecir el pasado. En series de tiempo los pliegues deben respetar el orden:

- Se entrena con el pasado y se valida con el **bloque inmediatamente siguiente**.
- En el pliegue siguiente, la ventana de entrenamiento **crece** e incluye el bloque que antes se usó para validar.

**En esta sección solo hay dos conjuntos:**

| Conjunto | Filas | Uso |
|---|---|---|
| Entrenamiento (70%) | 868 | Validación cruzada secuencial para elegir `max_depth` |
| Prueba = validación + prueba (30%) | 373 | Evaluar **una sola vez** el modelo final |

Ya no hace falta un conjunto de validación separado: la validación ocurre **dentro** del entrenamiento.

```python
from sklearn.model_selection import TimeSeriesSplit

X_prueba_cv, y_prueba_cv = X.iloc[n_train:], y.iloc[n_train:]     # validación + prueba
print(f"Entrenamiento: {len(X_train)} filas ({X_train.index[0].date()} a {X_train.index[-1].date()})")
print(f"Prueba:        {len(X_prueba_cv)} filas ({X_prueba_cv.index[0].date()} a {X_prueba_cv.index[-1].date()})")
```

Usamos `TimeSeriesSplit` de scikit-learn con **5 pliegues de 100 días** sobre las 868 filas de entrenamiento. El primer entrenamiento tiene $868 - 5 \times 100 = 368$ filas, aproximadamente un año.

```python
N_PLIEGUES, TAM_PLIEGUE = 5, 100
cv_secuencial = TimeSeriesSplit(n_splits=N_PLIEGUES, test_size=TAM_PLIEGUE)
pliegues = list(cv_secuencial.split(X_train))

pd.DataFrame([{"pliegue": k + 1,
               "filas entrenamiento": len(i_tr),
               "entrena hasta": X_train.index[i_tr[-1]].date(),
               "valida desde": X_train.index[i_va[0]].date(),
               "valida hasta": X_train.index[i_va[-1]].date()}
              for k, (i_tr, i_va) in enumerate(pliegues)]).set_index("pliegue")
```

```python
fig, ax = plt.subplots(figsize=(11, 3.5))
for k, (i_tr, i_va) in enumerate(pliegues):
    fila = N_PLIEGUES - k
    ax.barh(fila, X_train.index[i_tr[-1]] - X_train.index[i_tr[0]], left=X_train.index[i_tr[0]], color="tab:blue",
            label="entrenamiento" if k == 0 else None)
    ax.barh(fila, X_train.index[i_va[-1]] - X_train.index[i_va[0]], left=X_train.index[i_va[0]], color="tab:orange",
            label="validación" if k == 0 else None)
ax.barh(0, X_prueba_cv.index[-1] - X_prueba_cv.index[0], left=X_prueba_cv.index[0], color="tab:green",
        label="prueba (no se usa en la validación cruzada)")
ax.set_yticks(range(N_PLIEGUES + 1), ["prueba"] + [f"pliegue {N_PLIEGUES - j}" for j in range(N_PLIEGUES)])
ax.set_title("Validación cruzada secuencial con ventana creciente")
ax.legend(loc="lower left", fontsize=8); plt.show()
```

Para cada profundidad entrenamos un árbol en cada pliegue, calculamos el error en su bloque de validación y promediamos.

**Criterio:** promediamos el **ECM** de los pliegues y reportamos su raíz (RECM). Como todos los bloques tienen 100 días, el promedio de los ECM es igual al ECM de las 500 predicciones juntas; así lo calcula también `skforecast`, lo que permite comparar ambos resultados.

```python
from sklearn.metrics import mean_squared_error

ecm_pliegues = {}
for d in range(1, 16):
    ecm_pliegues[d] = []
    for i_tr, i_va in pliegues:
        arbol_d = DecisionTreeRegressor(max_depth=d, random_state=SEMILLA).fit(X_train.iloc[i_tr], y_train.iloc[i_tr])
        ecm_pliegues[d].append(mean_squared_error(y_train.iloc[i_va], arbol_d.predict(X_train.iloc[i_va])))
ecm_pliegues = pd.DataFrame(ecm_pliegues, index=[f"pliegue {k + 1}" for k in range(N_PLIEGUES)]).T
ecm_pliegues.index.name = "max_depth"

resultados_cv = np.sqrt(ecm_pliegues)                    # RECM de cada pliegue
resultados_cv["RECM_cv"] = np.sqrt(ecm_pliegues.mean(axis=1))
MEJOR_PROF_CV = resultados_cv["RECM_cv"].idxmin()
print("Mejor max_depth según validación cruzada:", MEJOR_PROF_CV, "| según la validación simple (4.3):", MEJOR_PROF)
resultados_cv.round(2)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
for col in resultados_cv.columns[:-1]:
    ax.plot(resultados_cv.index, resultados_cv[col], color="gray", alpha=0.5, lw=1)
ax.plot([], [], color="gray", alpha=0.5, label="cada pliegue")
ax.plot(resultados_cv.index, resultados_cv["RECM_cv"], color="tab:red", marker="o", lw=2.5, label="validación cruzada (promedio)")
ax.plot(resultados_prof.index, resultados_prof["RECM_validación"], color="tab:orange", marker="s", ls="--",
        label="validación simple (sección 4.3)")
ax.axvline(MEJOR_PROF_CV, color="tab:red", ls=":")
ax.axvline(MEJOR_PROF, color="tab:orange", ls=":")
ax.set_xlabel("max_depth"); ax.set_ylabel("RECM"); ax.set_title("Validación cruzada secuencial vs. validación simple")
ax.legend(fontsize=8); plt.show()
```

### El mismo cálculo con `skforecast`

En `skforecast` la validación cruzada secuencial se hace con `TimeSeriesFold`:

- `steps=1`: cada predicción es a un día, con los rezagos reales (igual que a mano).
- `initial_train_size = VENTANA + 368`: la serie necesaria para formar las 368 filas del primer pliegue.
- `refit=100`: el modelo se **reentrena cada 100 días**, que equivale a los 5 pliegues de 100 días.
- `fixed_train_size=False`: la ventana de entrenamiento **crece** en cada reentrenamiento.

```python
cv_secuencial_sk = TimeSeriesFold(
    steps=1,
    initial_train_size=VENTANA + len(pliegues[0][0]),
    refit=TAM_PLIEGUE,
    fixed_train_size=False,
)
resultados_cv_sk = grid_search_forecaster(
    forecaster=ForecasterRecursive(estimator=DecisionTreeRegressor(random_state=SEMILLA), lags=LAGS),
    y=pm25.iloc[:VENTANA + n_train],                 # solo la serie de entrenamiento
    cv=cv_secuencial_sk,
    param_grid={"max_depth": list(range(1, 16))},
    metric="mean_squared_error",
    return_best=False,
    n_jobs=1,
    show_progress=False,
)
ecm_sk = resultados_cv_sk.set_index("max_depth")["mean_squared_error"].sort_index()

print("Mismo ECM para cada profundidad:", np.allclose(ecm_sk.values, ecm_pliegues.mean(axis=1).values))
print("Misma profundidad elegida:      ", resultados_cv_sk.iloc[0]["max_depth"] == MEJOR_PROF_CV)
```

### Modelo final y evaluación en prueba

Con la profundidad elegida, entrenamos con **todo el entrenamiento** (868 filas) y evaluamos una sola vez en la prueba (373 días). Para comparar de forma justa, entrenamos también la profundidad elegida en la sección 4.3 con los **mismos datos** y la evaluamos en la **misma prueba**.

```python
arbol_cv = DecisionTreeRegressor(max_depth=MEJOR_PROF_CV, random_state=SEMILLA).fit(X_train, y_train)
pred_prueba_cv = arbol_cv.predict(X_prueba_cv)
arbol_43 = DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA).fit(X_train, y_train)

# Verificación con skforecast: backtesting sobre los 373 días de prueba
_, pred_prueba_cv_sk = backtesting_forecaster(
    forecaster=ForecasterRecursive(estimator=DecisionTreeRegressor(max_depth=MEJOR_PROF_CV, random_state=SEMILLA), lags=LAGS),
    y=pm25,
    cv=TimeSeriesFold(steps=1, initial_train_size=VENTANA + n_train, refit=False),
    metric="mean_squared_error",
    n_jobs=1,
    show_progress=False,
)
print("Predicciones de prueba iguales a las de skforecast:", np.allclose(pred_prueba_cv_sk["pred"].values, pred_prueba_cv))

pd.DataFrame({f"max_depth={MEJOR_PROF_CV} (validación cruzada)": {**evaluar(y_prueba_cv, pred_prueba_cv), "RECM estimada al elegir": resultados_cv.loc[MEJOR_PROF_CV, "RECM_cv"]},
              f"max_depth={MEJOR_PROF} (validación simple)":    {**evaluar(y_prueba_cv, arbol_43.predict(X_prueba_cv)), "RECM estimada al elegir": resultados_prof.loc[MEJOR_PROF, "RECM_validación"]},
              "ingenuo (y_t = y_(t-1))":                           {**evaluar(y_prueba_cv, X_prueba_cv["lag_1"]), "RECM estimada al elegir": np.nan}}).T.round(3)
```

**¿Qué aprendimos?**

1. **El error cambia mucho según la época del año.** Los pliegues que validan en invierno (enero–abril de 2015 y noviembre de 2015–febrero de 2016) tienen errores cercanos al doble que los de verano y otoño. Una sola ventana de validación ve solo una parte de esa variación.
2. **La validación simple fue optimista.** Su bloque cayó en verano, con poca contaminación, y estimó una RECM cercana a 14.6, muy por debajo del error real en la prueba (que incluye un invierno completo). La validación cruzada, al promediar varias épocas del año, da una estimación **más prudente** (cercana a 28), que en este caso queda por encima del error en prueba. Ninguna estimación es exacta, pero la de la validación cruzada no depende de una sola época del año.
3. **La curva es plana entre profundidades pequeñas.** Las RECM de `max_depth` entre 1 y 5 difieren muy poco frente a la variación entre pliegues. La validación cruzada elige un árbol más simple; en esta prueba la profundidad de la sección 4.3 resultó levemente mejor, pero la diferencia es pequeña y ambos superan al modelo ingenuo. Cuando varias opciones empatan en la práctica, es razonable preferir la más simple.
4. **Manual = `skforecast`.** `TimeSeriesSplit` con 5 pliegues de 100 días equivale a `TimeSeriesFold(steps=1, refit=100, fixed_train_size=False)`.

## 4.5 Alternativa: poda por costo-complejidad

En vez de limitar la profundidad, se puede **hacer crecer el árbol completo y luego podarlo**. La poda minimiza

$$R_\alpha(T) = \text{ECM}(T) + \alpha \, |T|,$$

donde $|T|$ es el número de hojas. Con $\alpha = 0$ se obtiene el árbol completo; al aumentar $\alpha$, el árbol se hace más pequeño. `cost_complexity_pruning_path` calcula todos los valores de $\alpha$ en los que cambia el árbol. Es importante calcularla sobre el **árbol completo** (sin `max_depth`).

```python
ruta = arbol_completo.cost_complexity_pruning_path(X_train, y_train)
alphas = ruta.ccp_alphas[:-1]          # el último alpha deja el árbol con una sola hoja
print("Número de valores de alpha:", len(alphas))

recm_tr, recm_val, n_hojas = [], [], []
for a in alphas:
    arbol_a = DecisionTreeRegressor(ccp_alpha=a, random_state=SEMILLA).fit(X_train, y_train)
    recm_tr.append(root_mean_squared_error(y_train, arbol_a.predict(X_train)))
    recm_val.append(root_mean_squared_error(y_val, arbol_a.predict(X_val)))
    n_hojas.append(arbol_a.get_n_leaves())

MEJOR_ALPHA = alphas[np.argmin(recm_val)]
arbol_podado = DecisionTreeRegressor(ccp_alpha=MEJOR_ALPHA, random_state=SEMILLA).fit(X_train, y_train)
print(f"Mejor alpha: {MEJOR_ALPHA:.3f} | RECM validación: {min(recm_val):.3f} | "
      f"profundidad: {arbol_podado.get_depth()} | hojas: {arbol_podado.get_n_leaves()}")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(alphas, recm_tr, label="entrenamiento")
ax[0].plot(alphas, recm_val, label="validación")
ax[0].axvline(MEJOR_ALPHA, color="red", ls="--", label=f"alpha = {MEJOR_ALPHA:.2f}")
ax[0].set_xscale("symlog", linthresh=0.1); ax[0].set_xlim(left=0); ax[0].set_xlabel("alpha (escala symlog)"); ax[0].set_ylabel("RECM")
ax[0].set_title("Error según alpha"); ax[0].legend()
ax[1].plot(alphas, n_hojas)
ax[1].set_xscale("symlog", linthresh=0.1); ax[1].set_xlim(left=0); ax[1].set_yscale("log")
ax[1].set_xlabel("alpha (escala symlog)"); ax[1].set_ylabel("número de hojas")
ax[1].set_title("Tamaño del árbol según alpha")
plt.tight_layout(); plt.show()
```

```python
arbol_prof = DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA).fit(X_train, y_train)
pd.DataFrame({f"max_depth={MEJOR_PROF}": {**evaluar(y_val, arbol_prof.predict(X_val)), "hojas": arbol_prof.get_n_leaves()},
              f"poda alpha={MEJOR_ALPHA:.2f}": {**evaluar(y_val, arbol_podado.predict(X_val)), "hojas": arbol_podado.get_n_leaves()}}
            ).round(3)
```

**Decisión.** Ambos métodos tienen un error de validación muy parecido. El árbol con `max_depth` tiene menos hojas y, sobre todo, es **mucho menos profundo**, así que sus reglas son más cortas y fáciles de explicar. Ante errores similares preferimos el modelo más simple, así que continuamos con el árbol de `max_depth` elegido. El árbol podado se incluye en la comparación final (sección 9).

## 4.6 Modelo final: entrenar con entrenamiento + validación y evaluar en prueba

Una vez elegido el hiperparámetro, **reentrenamos con entrenamiento + validación** (más datos, mejor modelo) y evaluamos **una única vez** en prueba.

```python
X_train_val = pd.concat([X_train, X_val])
y_train_val = pd.concat([y_train, y_val])

arbol_final = DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA).fit(X_train_val, y_train_val)
pred_test_manual = arbol_final.predict(X_test)

# Referencia ingenua: "mañana será igual que hoy" (predicción = lag_1)
pd.DataFrame({f"árbol max_depth={MEJOR_PROF}": evaluar(y_test, pred_test_manual),
              "ingenuo (y_t = y_(t-1))":        evaluar(y_test, X_test["lag_1"])}).round(3)
```

Siempre conviene comparar con un **modelo ingenuo**: si el árbol no lo supera, no vale la pena. Aquí el árbol mejora al ingenuo.

```python
pred_train_val = pd.Series(arbol_final.predict(X_train_val), index=y_train_val.index)
pred_test = pd.Series(pred_test_manual, index=y_test.index)

fig, ax = plt.subplots(figsize=(12, 4))
y.plot(ax=ax, color="gray", lw=1, label="observado")
pred_train_val.plot(ax=ax, color="tab:blue", lw=1, label="ajuste (entrenamiento + validación)")
pred_test.plot(ax=ax, color="tab:red", lw=1.2, label="predicción (prueba)")
ax.axvline(y_test.index[0], color="black", ls="--")
ax.set_title("Observado vs. predicho a 1 paso"); ax.set_ylabel("µg/m³"); ax.legend(fontsize=8)
plt.show()
```

Las predicciones toman pocos valores distintos (uno por hoja): es la función escalonada de la sección 4.1, ahora en varias dimensiones.

```python
plt.figure(figsize=(14, 6))
plot_tree(arbol_final, feature_names=X.columns, filled=True, rounded=True, fontsize=7, max_depth=2)
plt.title(f"Primeros niveles del árbol final (max_depth={MEJOR_PROF})")
plt.show()
```

# 5. El mismo proceso con `skforecast`

[`skforecast`](https://skforecast.org) automatiza lo que hicimos a mano: crea la tabla de rezagos, entrena y pronostica. Verificaremos **en cada paso** que los resultados son idénticos a los manuales.

`ForecasterRecursive` recibe la serie **original** (no la tabla) y la lista de rezagos.

## 5.1 Verificación 1: la tabla de rezagos

```python
forecaster = ForecasterRecursive(
    estimator=DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA),
    lags=LAGS,
)
X_sk, y_sk = forecaster.create_train_X_y(pm25)

print("Mismas columnas:   ", list(X_sk.columns) == list(X.columns))
print("Mismas fechas:     ", X_sk.index.equals(X.index))
print("Misma matriz X:    ", np.allclose(X_sk.values, X.values))
print("Misma respuesta y: ", np.allclose(y_sk.values, y.values))
```

## 5.2 Backtesting: evaluar como en la práctica

El **backtesting** simula el uso real del modelo: se entrena con los datos iniciales y luego se pronostica cada día, uno a la vez, usando los valores reales observados hasta el día anterior.

Parámetros de `TimeSeriesFold`:
- `initial_train_size`: cuántos datos de la **serie original** se usan para entrenar. Como los primeros `VENTANA` días solo sirven de rezagos, para entrenar con las `n_train` filas de la tabla se necesitan `VENTANA + n_train` días.
- `steps=1`: se pronostica un día adelante en cada partición.
- `refit=False`: el modelo se entrena **una sola vez**, igual que hicimos a mano.

```python
print(f"Días para entrenar = VENTANA + n_train = {VENTANA} + {n_train} = {VENTANA + n_train}")
print(f"Días hasta el final de validación = {VENTANA + n_train + n_val}")
```

## 5.3 Verificación 2: validación (selección de `max_depth`)

```python
cv_val = TimeSeriesFold(steps=1, initial_train_size=VENTANA + n_train, refit=False)

_, pred_val_sk = backtesting_forecaster(
    forecaster=forecaster,
    y=pm25.iloc[:VENTANA + n_train + n_val],     # la serie hasta el final de validación
    cv=cv_val,
    metric="mean_squared_error",
    n_jobs=1,
    show_progress=False,
)
print("Predicciones de validación iguales a las manuales:", np.allclose(pred_val_sk["pred"].values, arbol_prof.predict(X_val)))
```

## 5.4 Verificación 3: búsqueda en malla

`grid_search_forecaster` repite el backtesting de validación para cada combinación de rezagos e hiperparámetros. Incluimos nuestros rezagos (`"manual"`) junto con otras alternativas. Si todo es coherente, debe elegir la misma profundidad que encontramos a mano.

```python
forecaster_grid = ForecasterRecursive(estimator=DecisionTreeRegressor(random_state=SEMILLA), lags=LAGS)

resultados_grid = grid_search_forecaster(
    forecaster=forecaster_grid,
    y=pm25.iloc[:VENTANA + n_train + n_val],
    cv=cv_val,
    param_grid={"max_depth": list(range(1, 16))},
    lags_grid={"manual": LAGS, "1 a 7": 7, "1 a 14": 14, "1 a 30": 30},
    metric="mean_squared_error",
    return_best=False,
    n_jobs=1,
    show_progress=False,
)
resultados_grid["RECM"] = np.sqrt(resultados_grid["mean_squared_error"])
resultados_grid[["lags_label", "max_depth", "RECM"]].head(8)
```

```python
mejor = resultados_grid.iloc[0]
print("Mejores rezagos:", mejor["lags_label"], "| mejor max_depth:", mejor["max_depth"])
print("Coincide con la elección manual:", mejor["lags_label"] == "manual" and mejor["max_depth"] == MEJOR_PROF)
print(f"RECM de validación: skforecast {mejor['RECM']:.4f} vs manual {resultados_prof.loc[MEJOR_PROF, 'RECM_validación']:.4f}")
```

## 5.5 Verificación 4: modelo final en prueba

```python
cv_test = TimeSeriesFold(steps=1, initial_train_size=VENTANA + n_train + n_val, refit=False)

metrica_test_sk, pred_test_sk = backtesting_forecaster(
    forecaster=forecaster,
    y=pm25,
    cv=cv_test,
    metric="mean_squared_error",
    n_jobs=1,
    show_progress=False,
)
print("Predicciones de prueba iguales a las manuales:", np.allclose(pred_test_sk["pred"].values, pred_test_manual))
print(f"RECM prueba: skforecast {np.sqrt(metrica_test_sk.iloc[0, 0]):.4f} vs manual {evaluar(y_test, pred_test_manual)['RECM']:.4f}")
```

**Conclusión de la sección:** `skforecast` construye exactamente la misma tabla, entrena el mismo árbol y produce las mismas predicciones que el procedimiento manual. La ventaja es que lo hace con menos código y con menos oportunidades de equivocarse en los índices.

# 6. Explicabilidad

## 6.1 Importancia de variables

La importancia de una variable en un árbol es la **reducción total del error cuadrático** que logran las divisiones que usan esa variable (normalizada para que sume 1).

```python
forecaster.fit(y=pm25.iloc[:VENTANA + n_train + n_val])   # mismo entrenamiento que arbol_final
importancia = forecaster.get_feature_importances()

print("Importancias iguales a las del árbol manual:",
      np.allclose(importancia.set_index("feature").loc[X.columns, "importance"], arbol_final.feature_importances_))
importancia.head(10)
```

## 6.2 Valores SHAP

La importancia global dice *qué variables usa el árbol*, pero no *cómo* influyen en **cada predicción**. Los valores **SHAP** (SHapley Additive exPlanations) reparten cada predicción entre las variables:

$$\hat y_t \;=\; \underbrace{E[\hat y]}_{\text{valor base}} \;+\; \sum_{j} \phi_j(t)$$

- **Valor base** $E[\hat y]$: la predicción promedio del modelo; en un árbol es el promedio de la respuesta en entrenamiento.
- $\phi_j(t)$: cuánto **sube** (si es positivo) o **baja** (si es negativo) la variable $j$ la predicción del día $t$ respecto al valor base.
- La suma es **exacta**: valor base + suma de los SHAP = predicción.

Para árboles, `shap.TreeExplainer` calcula los valores exactos. Explicamos las predicciones del conjunto de **prueba**.

```python
explicador = shap.TreeExplainer(arbol_final)
shap_test = explicador(X_test)

print(f"Valor base: {explicador.expected_value[0]:.3f}  |  media de y en entrenamiento+validación: {y_train_val.mean():.3f}")
print("Valor base + suma de SHAP = predicción:",
      np.allclose(explicador.expected_value[0] + shap_test.values.sum(axis=1), pred_test_manual))
```

### Importancia global con SHAP

El gráfico de barras muestra el promedio de $|\phi_j|$: **cuántos µg/m³ mueve en promedio** cada variable la predicción.

```python
shap.plots.bar(shap_test, max_display=12)
```

### Gráfico de enjambre (*beeswarm*)

Cada punto es un día de prueba. La posición horizontal es el valor SHAP y el color es el **valor de la variable** (rojo = alto, azul = bajo).

**Cómo leerlo:** si los puntos rojos están a la derecha y los azules a la izquierda, valores altos de la variable **aumentan** la predicción. Las variables que el árbol no usa tienen todos sus puntos en cero.

```python
shap.plots.beeswarm(shap_test, max_display=12)
```

### Dependencia: cómo actúa `lag_1`

Este gráfico muestra el valor SHAP de `lag_1` frente al valor de `lag_1`. La forma **escalonada** es la huella del árbol: el efecto cambia de golpe en los puntos de corte del árbol.

```python
shap.plots.scatter(shap_test[:, "lag_1"], color=shap_test)
```

### Explicación de un día concreto (*waterfall*)

Elegimos el día de prueba con **menor** PM2.5 observado. El gráfico parte del valor base $E[f(X)]$ (abajo) y suma, variable por variable, las contribuciones hasta llegar a la predicción $f(x)$ (arriba). Las barras azules bajan la predicción y las rojas la suben.

```python
i_min = int(np.argmin(y_test.values))
print(f"Fecha: {y_test.index[i_min].date()} | observado: {y_test.iloc[i_min]:.1f} | predicho: {pred_test_manual[i_min]:.1f}")
shap.plots.waterfall(shap_test[i_min], max_display=10)
```

**Interpretación de los valores SHAP en este modelo:**

1. **`lag_1` domina.** El nivel de contaminación de ayer es la variable que más mueve la predicción: si ayer fue alto (puntos rojos), la predicción sube; si fue bajo, baja. Esto confirma lo que mostró la PACF.
2. **Los rezagos estacionales aportan.** Algunas variables `lag_364`–`lag_370` tienen contribuciones no nulas: el nivel de hace un año ayuda a ubicar la **época del año** (invierno con más contaminación, verano con menos), como sugería el periodograma.
3. **Muchas variables valen cero.** Un árbol de poca profundidad usa pocas variables; las que no aparecen en ninguna división tienen SHAP igual a cero para todos los días.
4. **Límite del modelo.** Como el árbol predice promedios de hojas, no alcanza los valores extremos (sección 4.1): en los días de contaminación muy alta la predicción queda muy por debajo del valor observado (ver ejercicio 6).
5. **SHAP explica al modelo, no a la realidad.** Una contribución grande indica que *el árbol* usa esa variable, no que exista una relación causal.

# 7. Pronóstico a varios pasos: recursivo vs. directo

Hasta ahora siempre conocíamos el valor real de ayer. Para pronosticar **varios días adelante** hay dos estrategias:

| Paso | **Recursiva** (un solo modelo) | **Directa** (un modelo por horizonte) |
|---|---|---|
| $t+1$ | modelo con $y_t, y_{t-1}, \dots$ | modelo$_1$ con $y_t, y_{t-1}, \dots$ |
| $t+2$ | mismo modelo con $\color{red}{\hat y_{t+1}}, y_t, \dots$ | modelo$_2$ con $y_t, y_{t-1}, \dots$ |
| $t+3$ | mismo modelo con $\color{red}{\hat y_{t+2}}, \color{red}{\hat y_{t+1}}, \dots$ | modelo$_3$ con $y_t, y_{t-1}, \dots$ |

- **Recursiva** (`ForecasterRecursive`): usa sus propias predicciones (en rojo) como rezagos. Los errores se acumulan.
- **Directa** (`ForecasterDirect`): entrena un modelo distinto para cada horizonte $h$, con respuesta $y_{t+h}$. No reutiliza predicciones, pero requiere entrenar más modelos.

Comparamos ambas estrategias pronosticando los primeros 14 días del conjunto de prueba, con modelos entrenados con entrenamiento + validación.

```python
H = 14
y_hasta_val = pm25.iloc[:VENTANA + n_train + n_val]

forecaster.fit(y=y_hasta_val, store_in_sample_residuals=True)
pred_rec = forecaster.predict_interval(steps=H, interval=[0.025, 0.975], method="bootstrapping", n_boot=200)

forecaster_dir = ForecasterDirect(
    estimator=DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA),
    lags=LAGS, steps=H, n_jobs=1,
)
forecaster_dir.fit(y=y_hasta_val)
pred_dir = forecaster_dir.predict()

reales_H = pm25.loc[pred_rec.index]
pd.DataFrame({"recursivo": evaluar(reales_H, pred_rec["pred"]), "directo": evaluar(reales_H, pred_dir)}).round(3)
```

```python
fig, ax = plt.subplots(figsize=(11, 4))
pm25.loc[pred_rec.index[0] - pd.Timedelta(days=20):pred_rec.index[-1]].plot(ax=ax, color="black", marker=".", label="observado")
pred_rec["pred"].plot(ax=ax, color="tab:red", marker="o", label="recursivo")
ax.fill_between(pred_rec.index, pred_rec["lower_bound"], pred_rec["upper_bound"], color="tab:red", alpha=0.15,
                label="intervalo 95% (recursivo, bootstrap)")
pred_dir.plot(ax=ax, color="tab:blue", marker="s", label="directo")
ax.axvline(pred_rec.index[0], color="gray", ls="--")
ax.set_title(f"Pronóstico a {H} días"); ax.set_ylabel("µg/m³"); ax.legend(fontsize=8)
plt.show()
```

**¿Qué observamos?**

- El pronóstico **recursivo se aplana** rápidamente. Al alimentar el árbol con sus propias predicciones, que son promedios de hojas, las entradas convergen a valores "típicos" y el árbol cae una y otra vez en la misma hoja.
- El pronóstico **directo** no se aplana de la misma forma, porque cada horizonte tiene su propio modelo que usa solo valores observados.
- En una ventana de solo 14 días el R² no es informativo (la varianza de los datos reales es pequeña); conviene fijarse en la RECM.
- Los intervalos son **anchos**: pronosticar varios días adelante con un modelo tan simple tiene mucha incertidumbre.

# 8. Etapa de producción: pronóstico 1 paso adelante

Ya elegimos el modelo (rezagos y `max_depth`) y medimos su desempeño en prueba. Para **usarlo en la práctica**:

1. Se reentrena con **todas las observaciones disponibles** (entrenamiento + validación + prueba): los datos más recientes son los más valiosos.
2. Se construye la fila de entrada con las **últimas observaciones** de la muestra.
3. Se pronostica el **día siguiente** al último dato, que todavía no se ha observado.

> En producción no hay un valor real con el cual comparar: la evaluación ya se hizo en la secciones 4.6 y 5.5.

## 8.1 A mano

```python
arbol_produccion = DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA).fit(X, y)   # todas las filas

ultima_fecha = pm25.index[-1]
fecha_pronostico = ultima_fecha + pd.Timedelta(days=1)
print(f"Último dato observado: {ultima_fecha.date()}  ->  pronosticamos: {fecha_pronostico.date()}")

# Fila de entrada: el rezago k del día a pronosticar es el valor observado k días antes de esa fecha
x_nuevo = pd.DataFrame([[pm25[fecha_pronostico - pd.Timedelta(days=k)] for k in LAGS]], columns=X.columns, index=[fecha_pronostico])

entrada = pd.DataFrame({"fecha del dato": [(fecha_pronostico - pd.Timedelta(days=k)).date() for k in LAGS],
                        "valor": x_nuevo.iloc[0].values}, index=X.columns)
entrada
```

```python
pred_produccion_manual = arbol_produccion.predict(x_nuevo)[0]
print(f"Pronóstico manual de PM2.5 para {fecha_pronostico.date()}: {pred_produccion_manual:.3f} µg/m³")
```

## 8.2 Con `skforecast`

`fit` guarda internamente la **última ventana** (`last_window_`): los últimos `VENTANA` valores, que son los necesarios para formar la fila del día siguiente. `create_predict_X` muestra esa fila y `predict` produce el pronóstico.

```python
forecaster_produccion = ForecasterRecursive(
    estimator=DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA),
    lags=LAGS,
)
forecaster_produccion.fit(y=pm25, store_in_sample_residuals=True)

print("Tamaño de la última ventana:", len(forecaster_produccion.last_window_))
x_nuevo_sk = forecaster_produccion.create_predict_X(steps=1)
print("Misma fila de entrada que la manual:", np.allclose(x_nuevo_sk.values, x_nuevo.values))

pred_produccion_sk = forecaster_produccion.predict(steps=1)
print("Mismo pronóstico que el manual:     ", np.isclose(pred_produccion_sk.iloc[0], pred_produccion_manual))
pred_produccion_sk
```

```python
intervalo_produccion = forecaster_produccion.predict_interval(steps=1, interval=[0.025, 0.975], method="bootstrapping", n_boot=500)
intervalo_produccion
```

```python
fig, ax = plt.subplots(figsize=(11, 4))
ultimos = pm25.iloc[-60:]
ip = intervalo_produccion.iloc[0]
ax.plot(ultimos.index, ultimos.values, color="black", marker=".", label="últimas observaciones")
ax.errorbar([fecha_pronostico], [ip["pred"]],
            yerr=[[ip["pred"] - ip["lower_bound"]], [ip["upper_bound"] - ip["pred"]]],
            fmt="o", color="tab:red", capsize=5, label="pronóstico e intervalo 95%")
ax.set_title(f"Pronóstico de producción para {fecha_pronostico.date()}"); ax.set_ylabel("µg/m³"); ax.legend()
plt.show()
```

## 8.3 ¿Por qué el modelo predice ese valor?

Con SHAP explicamos el pronóstico de producción, igual que en la sección 6.

```python
explicador_prod = shap.TreeExplainer(arbol_produccion)
shap.plots.waterfall(explicador_prod(x_nuevo)[0], max_display=10)
```

# 9. Comparación final y conclusiones

Todos los modelos se entrenan con entrenamiento + validación y se evalúan en prueba, pronosticando 1 paso adelante.

```python
modelos = {
    "Ingenuo (y_t = y_(t-1))": None,
    "Árbol completo": DecisionTreeRegressor(random_state=SEMILLA),
    f"Árbol podado (alpha={MEJOR_ALPHA:.2f})": DecisionTreeRegressor(ccp_alpha=MEJOR_ALPHA, random_state=SEMILLA),
    f"Árbol max_depth={MEJOR_PROF} (manual = skforecast)": DecisionTreeRegressor(max_depth=MEJOR_PROF, random_state=SEMILLA),
}
comparacion = {}
for nombre, modelo in modelos.items():
    pred = X_test["lag_1"] if modelo is None else modelo.fit(X_train_val, y_train_val).predict(X_test)
    comparacion[nombre] = evaluar(y_test, pred)
comparacion = pd.DataFrame(comparacion).T.sort_values("RECM")
comparacion.round(3)
```

```python
ax = comparacion["RECM"].sort_values(ascending=False).plot.barh(color="tab:blue")
ax.bar_label(ax.containers[0], fmt="%.2f", padding=3)
ax.set_xlabel("RECM en prueba (µg/m³)"); ax.set_title("Comparación de modelos (menor es mejor)")
ax.set_xlim(0, comparacion["RECM"].max() * 1.15)
plt.show()
```

## Conclusiones

1. **Diagnóstico → variables.** La PACF (día anterior) y el periodograma (ciclo anual) justificaron los rezagos 1–10 y 364–370; los valores SHAP confirmaron que el modelo usa precisamente esas dos fuentes de información.
2. **Controlar la complejidad es esencial.** El árbol completo memoriza el entrenamiento y predice peor que el modelo ingenuo; limitar la profundidad (o podar) corrige el sobreajuste. La elección se hace con **validación**, y la **prueba se usa una sola vez**.
3. **Manual = `skforecast`.** La tabla de rezagos, la validación, la búsqueda en malla, las predicciones de prueba, la importancia de variables y el pronóstico de producción coinciden exactamente.
4. **Limitaciones de los árboles.** Predicen con escalones (promedios de hojas), no extrapolan, subestiman los valores extremos y, con la estrategia recursiva, el pronóstico a varios pasos se aplana.
5. **Siguientes pasos.** Bosques aleatorios y *gradient boosting* combinan muchos árboles y suavizan los escalones; quitar la tendencia permite usar árboles en series no estacionarias.
